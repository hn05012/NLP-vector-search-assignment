```python
from sentence_transformers import SentenceTransformer, util
import numpy as np

# Load the model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Sample texts
texts = [
    "The sky is blue.",
    "The sun is bright today.",
    "It's a sunny day.",
    "Rainy weather is gloomy.",
    "I enjoy watching the sunrise.",
    "He went for a walk in the park.",
    "The grass is green.",
    "Stars twinkle in the night sky.",
    "Clouds are fluffy today.",
    "Thunderstorms are loud."
]

# Encode texts
embeddings = model.encode(texts, convert_to_tensor=True)

# Compute cosine similarities
cosine_similarities = util.pytorch_cos_sim(embeddings, embeddings)
print("Cosine Similarity Matrix:\n", cosine_similarities)
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Convert to numpy for sklearn
embeddings_np = embeddings.cpu().numpy()

# Reduce dimensions
pca = PCA(n_components=2)
reduced = pca.fit_transform(embeddings_np)

# Plot
plt.figure(figsize=(10, 6))
for i, text in enumerate(texts):
    plt.scatter(reduced[i, 0], reduced[i, 1])
    plt.annotate(text, (reduced[i, 0], reduced[i, 1]))
plt.title("PCA of Sentence Embeddings")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.grid(True)
plt.show()
```

```python
import faiss

# Ensure float32 for FAISS
embedding_dim = embeddings_np.shape[1]
index = faiss.IndexFlatL2(embedding_dim)

# Normalize for cosine similarity (optional)
faiss.normalize_L2(embeddings_np)

# Add embeddings to the index
index.add(embeddings_np)

# Query with a variation of an existing sentence
query = "Today is bright and sunny."
query_embedding = model.encode([query])
faiss.normalize_L2(query_embedding)

# Search for similar texts
k = 3
distances, indices = index.search(query_embedding, k)
print("\nQuery:", query)
print("Top matches:")
for idx in indices[0]:
    print("-", texts[idx])
```
